```python
dataset = load_dataset("dair-ai/emotion")

train_df = dataset["train"].to_pandas()
test_df = dataset["test"].to_pandas()

train_df.to_csv(PROCESSED_DIR / "emotion_train.csv", index=False)
test_df.to_csv(PROCESSED_DIR / "emotion_test.csv", index=False)

print("Training shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Columns:", train_df.columns.tolist())
print("Emotion labels:", dataset["train"].features["label"].names)

display(train_df.head())
```

```python
train_df = train_df.dropna(subset=["text", "label"]).copy()
test_df = test_df.dropna(subset=["text", "label"]).copy()

train_df["text"] = train_df["text"].astype(str).str.strip()
test_df["text"] = test_df["text"].astype(str).str.strip()

train_df = train_df[train_df["text"] != ""].reset_index(drop=True)
test_df = test_df[test_df["text"] != ""].reset_index(drop=True)

print("Training samples:", len(train_df))
print("Test samples:", len(test_df))
print("Training class distribution:")
print(train_df["label"].value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split

train_texts, val_texts, train_labels, val_labels = train_test_split(
    train_df["text"].tolist(),
    train_df["label"].tolist(),
    test_size=0.15,
    random_state=42,
    stratify=train_df["label"]
)

num_labels = len(dataset["train"].features["label"].names)

print("Training:", len(train_texts))
print("Validation:", len(val_texts))
print("Number of classes:", num_labels)
```

```python
import torch
from transformers import DistilBertTokenizer, DistilBertForSequenceClassification

model_name = "distilbert-base-uncased"

tokenizer = DistilBertTokenizer.from_pretrained(model_name)

model = DistilBertForSequenceClassification.from_pretrained(
    model_name,
    num_labels=num_labels
)

device = torch.device(
    "mps" if torch.backends.mps.is_available()
    else "cuda" if torch.cuda.is_available()
    else "cpu"
)

model.to(device)

print("Model loaded on:", device)
```

```python
from torch.utils.data import Dataset, DataLoader

class EmotionDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_length=128):
        self.encodings = tokenizer(
            texts,
            truncation=True,
            padding="max_length",
            max_length=max_length,
            return_tensors="pt"
        )
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, index):
        return {
            "input_ids": self.encodings["input_ids"][index],
            "attention_mask": self.encodings["attention_mask"][index],
            "labels": self.labels[index]
        }

train_dataset = EmotionDataset(train_texts, train_labels, tokenizer)
val_dataset = EmotionDataset(val_texts, val_labels, tokenizer)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=16)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

```python
from datasets import load_dataset

emotion_data = load_dataset("dair-ai/emotion")
label_names = emotion_data["train"].features["label"].names

train_df = emotion_data["train"].to_pandas()
val_df = emotion_data["validation"].to_pandas()
test_df = emotion_data["test"].to_pandas()

def convert_examples(df):
    examples = []
    for row in df.itertuples(index=False):
        emotion = label_names[int(row.label)]
        examples.append({
            "messages": [
                {
                    "role": "user",
                    "content": f"Classify the emotion in this text. Return only one of {', '.join(label_names)}.\nText: {row.text}"
                },
                {
                    "role": "assistant",
                    "content": emotion
                }
            ]
        })
    return examples

for filename, frame in [
    ("train.jsonl", train_df),
    ("valid.jsonl", val_df),
    ("test.jsonl", test_df)
]:
    with open(DATA_DIR / filename, "w", encoding="utf-8") as file:
        for item in convert_examples(frame):
            file.write(json.dumps(item, ensure_ascii=False) + "\n")

print("Training files created:", DATA_DIR)
```

```python
from pathlib import Path
import re
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from mlx_lm import load, generate
from tqdm.auto import tqdm

BASE_DIR = Path.cwd()
TEST_PATH = BASE_DIR / "data" / "processed" / "emotion_test.csv"
ADAPTER_PATH = BASE_DIR.parent / "models" / "finetuned" / "qwen3-8b" / "adapters"

test_df = pd.read_csv(TEST_PATH)
model, tokenizer = load(
    "mlx-community/Qwen3-8B-4bit",
    adapter_path=str(ADAPTER_PATH)
)

label_names = ["sadness", "joy", "love", "anger", "fear", "surprise"]

def predict_emotion(text):
    prompt = (
        "Classify the emotion in the text into exactly one label: "
        "sadness, joy, love, anger, fear, surprise.\n"
        f"Text: {text}\n"
        "Emotion:"
    )
    output = generate(
        model,
        tokenizer,
        prompt=prompt,
        max_tokens=20,
        verbose=False
    )
    match = re.search(
        r"\b(sadness|joy|love|anger|fear|surprise)\b",
        output.lower()
    )
    return match.group(1) if match else "unknown"

tqdm.pandas()
test_df["predicted_emotion"] = test_df["text"].progress_apply(predict_emotion)
test_df["actual_emotion"] = test_df["label"].apply(lambda x: label_names[int(x)])

print("Accuracy:", accuracy_score(test_df["actual_emotion"], test_df["predicted_emotion"]))
print(classification_report(
    test_df["actual_emotion"],
    test_df["predicted_emotion"],
    labels=label_names,
    zero_division=0
))
```

```python
from tqdm.auto import tqdm
import re

def predict_with_model(model, tokenizer, text):
    prompt = (
        "Classify the emotion into exactly one label: "
        "sadness, joy, love, anger, fear, surprise.\n"
        f"Text: {text}\n"
        "Emotion:"
    )

    result = generate(
        model,
        tokenizer,
        prompt=prompt,
        max_tokens=80,
        verbose=False
    )

    result = result.split("</think>")[-1].strip().lower()
    match = re.search(
        r"\b(sadness|joy|love|anger|fear|surprise)\b",
        result
    )

    return match.group(1) if match else "unknown"

comparison_df = test_df.head(100).copy()

comparison_df["base_prediction"] = [
    predict_with_model(base_model, base_tokenizer, text)
    for text in tqdm(comparison_df["text"], desc="Base model")
]

comparison_df["finetuned_prediction"] = [
    predict_with_model(model, tokenizer, text)
    for text in tqdm(comparison_df["text"], desc="Fine-tuned model")
]

comparison_df["actual_emotion"] = comparison_df["label"].map(
    lambda x: label_names[int(x)]
)

print("Predictions completed.")
```

```python
from sklearn.metrics import accuracy_score, f1_score

results = pd.DataFrame([
    {
        "Model": "Base Qwen3-8B",
        "Accuracy": accuracy_score(
            comparison_df["actual_emotion"],
            comparison_df["base_prediction"]
        ),
        "Macro F1": f1_score(
            comparison_df["actual_emotion"],
            comparison_df["base_prediction"],
            labels=label_names,
            average="macro",
            zero_division=0
        )
    },
    {
        "Model": "Fine-tuned Qwen3-8B",
        "Accuracy": accuracy_score(
            comparison_df["actual_emotion"],
            comparison_df["finetuned_prediction"]
        ),
        "Macro F1": f1_score(
            comparison_df["actual_emotion"],
            comparison_df["finetuned_prediction"],
            labels=label_names,
            average="macro",
            zero_division=0
        )
    }
])

display(results.round(4))
```

```python
import re
import json
from mlx_lm import generate

def generate_marketing_insight(experiment_text):
    prompt = f"""You are CausalMark, a marketing experiment analysis assistant.

Analyze the experiment details below.
Identify the likely business impact, possible risks, and recommended next action.

Return valid JSON with these keys:
summary, business_impact, risks, recommendation.

Experiment:
{experiment_text}

JSON:"""

    output = generate(
        model,
        tokenizer,
        prompt=prompt,
        max_tokens=250,
        verbose=False
    )

    output = output.split("</think>")[-1].strip()
    match = re.search(r"\{.*\}", output, re.DOTALL)

    if match:
        try:
            return json.loads(match.group())
        except json.JSONDecodeError:
            pass

    return {
        "summary": output,
        "business_impact": "Requires further analysis",
        "risks": "Output could not be parsed as JSON",
        "recommendation": "Review the experiment data"
    }
```

```python
experiment = """
Campaign: Discount email campaign
Control conversion rate: 8.0%
Treatment conversion rate: 10.5%
Sample size: 5000 users per group
Treatment cost: 2000 currency units
Business objective: Increase conversions profitably
"""

insight = generate_marketing_insight(experiment)
print(json.dumps(insight, indent=4))
```

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

def analyze_ab_test(control_conversions, control_users,
                    treatment_conversions, treatment_users):

    p_control = control_conversions / control_users
    p_treatment = treatment_conversions / treatment_users

    pooled_rate = (
        control_conversions + treatment_conversions
    ) / (control_users + treatment_users)

    standard_error = np.sqrt(
        pooled_rate * (1 - pooled_rate) *
        (1 / control_users + 1 / treatment_users)
    )

    z_score = (p_treatment - p_control) / standard_error
    p_value = 2 * norm.sf(abs(z_score))

    return {
        "control_rate": p_control,
        "treatment_rate": p_treatment,
        "absolute_lift": p_treatment - p_control,
        "relative_lift": (
            (p_treatment - p_control) / p_control
            if p_control else None
        ),
        "z_score": z_score,
        "p_value": p_value,
        "statistically_significant": p_value < 0.05
    }
```

```python
ab_result = analyze_ab_test(
    control_conversions=400,
    control_users=5000,
    treatment_conversions=525,
    treatment_users=5000
)

display(pd.DataFrame([ab_result]).round(4))
```

```python
result = ab_result

if result["statistically_significant"] and result["absolute_lift"] > 0:
    decision = "Treatment shows a statistically significant improvement."
elif result["statistically_significant"] and result["absolute_lift"] < 0:
    decision = "Treatment shows a statistically significant decline."
else:
    decision = "Insufficient evidence of a difference at the 5% significance level."

print(decision)
print(f"Control conversion rate: {result['control_rate']:.2%}")
print(f"Treatment conversion rate: {result['treatment_rate']:.2%}")
print(f"Absolute lift: {result['absolute_lift']:.2%}")
print(f"Relative lift: {result['relative_lift']:.2%}" if result["relative_lift"] is not None else "Relative lift: undefined")
print(f"P-value: {result['p_value']:.6f}")
```

```python
import json
import random
from pathlib import Path
from scipy.stats import norm

random.seed(42)

PROJECT_DIR = Path.cwd().parent
MARKETING_DATA_DIR = PROJECT_DIR / "data" / "marketing_instructions"
MARKETING_DATA_DIR.mkdir(parents=True, exist_ok=True)

examples = []

scenarios = [
    ("email campaign", "conversion rate"),
    ("discount campaign", "purchase rate"),
    ("landing page redesign", "signup rate"),
    ("push notification", "purchase rate"),
    ("personalized recommendation", "click-through rate"),
    ("retargeting campaign", "conversion rate"),
    ("loyalty campaign", "repeat purchase rate"),
    ("onboarding email", "activation rate"),
]

for i in range(1000):
    campaign, metric = random.choice(scenarios)

    control_users = random.choice([1000, 2000, 3000, 5000, 8000])
    treatment_users = random.choice([1000, 2000, 3000, 5000, 8000])

    control_rate = random.uniform(0.03, 0.20)
    treatment_rate = max(
        0.001,
        min(0.95, control_rate + random.uniform(-0.04, 0.06))
    )

    control_conversions = round(control_users * control_rate)
    treatment_conversions = round(treatment_users * treatment_rate)

    p1 = control_conversions / control_users
    p2 = treatment_conversions / treatment_users
    pooled = (
        control_conversions + treatment_conversions
    ) / (control_users + treatment_users)

    se = (
        pooled * (1 - pooled)
        * (1 / control_users + 1 / treatment_users)
    ) ** 0.5

    z = (p2 - p1) / se if se > 0 else 0
    p_value = float(2 * norm.sf(abs(z)))

    absolute_lift = p2 - p1
    relative_lift = absolute_lift / p1 if p1 else None
    significant = p_value < 0.05

    if significant and absolute_lift > 0:
        recommendation = "Consider a gradual rollout while monitoring guardrail metrics."
    elif significant and absolute_lift < 0:
        recommendation = "Do not roll out this treatment without further investigation."
    else:
        recommendation = "Gather more evidence or run a follow-up experiment."

    user_prompt = (
        f"Analyze this marketing A/B test for CausalMark. "
        f"Campaign: {campaign}. Primary metric: {metric}. "
        f"Control: {control_conversions}/{control_users} conversions. "
        f"Treatment: {treatment_conversions}/{treatment_users} conversions. "
        f"Explain the lift, statistical evidence, limitations, and next action."
    )

    answer = {
        "summary": (
            f"Control rate was {p1:.2%}; treatment rate was {p2:.2%}. "
            f"Absolute lift was {absolute_lift:.2%}."
        ),
        "relative_lift": (
            round(relative_lift, 4) if relative_lift is not None else None
        ),
        "p_value": round(p_value, 6),
        "statistically_significant": significant,
        "business_impact": (
            "The observed rate changed, but revenue impact cannot be determined "
            "without order value, margin, and campaign cost data."
        ),
        "limitations": [
            "Statistical significance does not establish profitability.",
            "The conclusion depends on valid randomization and independent observations.",
            "Multiple testing and experiment duration may affect interpretation."
        ],
        "recommendation": recommendation
    }

    examples.append({
        "messages": [
            {
                "role": "user",
                "content": user_prompt
            },
            {
                "role": "assistant",
                "content": json.dumps(answer, ensure_ascii=False)
            }
        ]
    })

output_path = MARKETING_DATA_DIR / "marketing_train_seed.jsonl"

with open(output_path, "w", encoding="utf-8") as f:
    for example in examples:
        f.write(json.dumps(example, ensure_ascii=False) + "\n")

print("Examples generated:", len(examples))
print("Saved to:", output_path)
```

```python
from scipy.stats import norm

group_results = (
    df_real.groupby("segment")
    .agg(
        customers=("segment", "size"),
        visits=("visit", "sum"),
        conversions=("conversion", "sum"),
        total_spend=("spend", "sum"),
        conversion_rate=("conversion", "mean"),
        average_spend=("spend", "mean")
    )
    .reset_index()
)

display(group_results.round(4))
```

```python
control_name = "No E-Mail"

if control_name not in df_real["segment"].values:
    raise ValueError(
        f"Control group '{control_name}' not found. "
        f"Available groups: {df_real['segment'].unique().tolist()}"
    )

control = df_real[df_real["segment"] == control_name]

control_n = len(control)
control_rate = control["conversion"].mean()
control_spend = control["spend"].mean()

lift_rows = []

for segment in df_real["segment"].unique():
    treatment = df_real[df_real["segment"] == segment]

    n_t = len(treatment)
    rate_t = treatment["conversion"].mean()
    spend_t = treatment["spend"].mean()

    if segment == control_name:
        continue

    pooled = (
        treatment["conversion"].sum() + control["conversion"].sum()
    ) / (n_t + control_n)

    se = np.sqrt(
        pooled * (1 - pooled) * (1 / n_t + 1 / control_n)
    )

    z = (rate_t - control_rate) / se if se > 0 else 0
    p_value = float(2 * norm.sf(abs(z)))

    lift_rows.append({
        "treatment": segment,
        "treatment_customers": n_t,
        "control_customers": control_n,
        "control_conversion_rate": control_rate,
        "treatment_conversion_rate": rate_t,
        "absolute_lift": rate_t - control_rate,
        "relative_lift": (
            (rate_t - control_rate) / control_rate
            if control_rate > 0 else np.nan
        ),
        "p_value": p_value,
        "significant_at_0.05": p_value < 0.05,
        "average_spend_control": control_spend,
        "average_spend_treatment": spend_t
    })

lift_df = pd.DataFrame(lift_rows)
display(lift_df.round(5))
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

CHUNK_SIZE = 250_000

required_cols = ["treatment", "conversion", "visit", "exposure"]
totals = {
    "rows": 0,
    "treatment_count": 0,
    "control_count": 0,
    "treatment_conversions": 0,
    "control_conversions": 0,
    "treatment_visits": 0,
    "control_visits": 0,
    "treatment_exposures": 0,
    "control_exposures": 0
}

for chunk in pd.read_csv(CRITEO_PATH, chunksize=CHUNK_SIZE):
    missing = set(required_cols) - set(chunk.columns)
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    treatment = chunk["treatment"].astype(int)
    conversion = chunk["conversion"].astype(int)
    visit = chunk["visit"].astype(int)
    exposure = chunk["exposure"].astype(int)

    treated = treatment == 1
    control = treatment == 0

    totals["rows"] += len(chunk)
    totals["treatment_count"] += int(treated.sum())
    totals["control_count"] += int(control.sum())
    totals["treatment_conversions"] += int(conversion[treated].sum())
    totals["control_conversions"] += int(conversion[control].sum())
    totals["treatment_visits"] += int(visit[treated].sum())
    totals["control_visits"] += int(visit[control].sum())
    totals["treatment_exposures"] += int(exposure[treated].sum())
    totals["control_exposures"] += int(exposure[control].sum())

print("Rows processed:", f"{totals['rows']:,}")
print("Treatment records:", f"{totals['treatment_count']:,}")
print("Control records:", f"{totals['control_count']:,}")
```

```python
def safe_rate(numerator, denominator):
    return numerator / denominator if denominator else np.nan

criteo_summary = pd.DataFrame([
    {
        "group": "Control",
        "customers": totals["control_count"],
        "conversion_rate": safe_rate(
            totals["control_conversions"], totals["control_count"]
        ),
        "visit_rate": safe_rate(
            totals["control_visits"], totals["control_count"]
        ),
        "exposure_rate": safe_rate(
            totals["control_exposures"], totals["control_count"]
        )
    },
    {
        "group": "Treatment",
        "customers": totals["treatment_count"],
        "conversion_rate": safe_rate(
            totals["treatment_conversions"], totals["treatment_count"]
        ),
        "visit_rate": safe_rate(
            totals["treatment_visits"], totals["treatment_count"]
        ),
        "exposure_rate": safe_rate(
            totals["treatment_exposures"], totals["treatment_count"]
        )
    }
])

display(criteo_summary.round(6))
```

```python
control_rate = criteo_summary.loc[
    criteo_summary["group"] == "Control", "conversion_rate"
].iloc[0]

treatment_rate = criteo_summary.loc[
    criteo_summary["group"] == "Treatment", "conversion_rate"
].iloc[0]

absolute_lift = treatment_rate - control_rate
relative_lift = absolute_lift / control_rate if control_rate > 0 else np.nan

print(f"Control conversion rate: {control_rate:.4%}")
print(f"Treatment conversion rate: {treatment_rate:.4%}")
print(f"Absolute lift: {absolute_lift:.4%}")
print(f"Relative lift: {relative_lift:.2%}")
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

SAMPLE_SIZE = 50_000
CHUNK_SIZE = 250_000

rng = np.random.default_rng(42)
sample_df = None
seen = 0

for chunk in pd.read_csv(
    CRITEO_PATH,
    compression="gzip",
    chunksize=CHUNK_SIZE
):
    if sample_df is None:
        sample_df = chunk.iloc[:0].copy()

    for _, row in chunk.iterrows():
        seen += 1

        if len(sample_df) < SAMPLE_SIZE:
            sample_df.loc[len(sample_df)] = row
        else:
            j = rng.integers(0, seen)
            if j < SAMPLE_SIZE:
                sample_df.iloc[j] = row

    print(f"Processed {seen:,} rows; sampled {len(sample_df):,}")

sample_df = sample_df.reset_index(drop=True)

print("Final sample:", sample_df.shape)
display(sample_df.head())
```

```python
import pandas as pd
import json
from pathlib import Path

OUTPUT_DIR = Path(MARKETING_DATA_DIR)
SAMPLE_PATH = OUTPUT_DIR / "criteo_sample_50000.csv"
TRAIN_PATH = OUTPUT_DIR / "train.jsonl"

df = pd.read_csv(SAMPLE_PATH)

required = {"treatment", "conversion", "visit", "exposure"}
assert required.issubset(df.columns), f"Missing columns: {required - set(df.columns)}"

df = df.dropna(subset=list(required)).reset_index(drop=True)

# Use actual records; don't invent individual uplift labels.
def make_example(row, idx):
    treatment = int(row["treatment"])
    conversion = int(row["conversion"])
    visit = int(row["visit"])
    exposure = int(row["exposure"])

    user_prompt = f"""
Analyze this observed marketing record.
Treatment assignment: {treatment}
Conversion observed: {conversion}
Visit observed: {visit}
Exposure observed: {exposure}

Explain what is known and what cannot be concluded about this individual.
""".strip()

    assistant_answer = (
        f"The record has treatment assignment {treatment}, "
        f"conversion outcome {conversion}, visit outcome {visit}, "
        f"and exposure outcome {exposure}. "
        "These are observed outcomes only. This single record does not "
        "show what would have happened to the same individual under the "
        "other treatment assignment, so individual causal uplift cannot "
        "be determined from it."
    )

    return {
        "messages": [
            {"role": "user", "content": user_prompt},
            {"role": "assistant", "content": assistant_answer}
        ]
    }

with open(TRAIN_PATH, "w", encoding="utf-8") as f:
    for idx, row in df.iterrows():
        example = make_example(row, idx)
        f.write(json.dumps(example, ensure_ascii=False) + "\n")

print("Saved:", TRAIN_PATH)
print("Training examples:", len(df))
```

```python
import pandas as pd
import numpy as np
import json
from pathlib import Path

df = pd.read_csv(SAMPLE_PATH).dropna().reset_index(drop=True)

FEATURES = [f"f{i}" for i in range(12)]
rng = np.random.default_rng(42)

def build_example(row, task_type):
    features = {f: round(float(row[f]), 4) for f in FEATURES}
    treatment = int(row["treatment"])
    conversion = int(row["conversion"])
    visit = int(row["visit"])
    exposure = int(row["exposure"])

    if task_type == 0:
        prompt = (
            f"Describe this marketing record: features={features}, "
            f"treatment={treatment}, conversion={conversion}, "
            f"visit={visit}, exposure={exposure}. "
            "Separate observed facts from unknown causal effects."
        )
        answer = (
            f"The observed treatment assignment is {treatment}; conversion is "
            f"{conversion}, visit is {visit}, and exposure is {exposure}. "
            f"The supplied features are {features}. These observations alone "
            "do not reveal this customer's counterfactual outcome or individual uplift."
        )

    elif task_type == 1:
        prompt = (
            f"Assess the evidence in this record for campaign effectiveness: "
            f"features={features}, treatment={treatment}, conversion={conversion}, "
            f"visit={visit}, exposure={exposure}. Can it prove the campaign worked?"
        )
        answer = (
            f"This record has treatment={treatment}, conversion={conversion}, "
            f"visit={visit}, and exposure={exposure}. One record cannot establish "
            "campaign effectiveness. Compare treatment and control groups using "
            "the experiment design, sample sizes, outcome rates, and uncertainty."
        )

    elif task_type == 2:
        prompt = (
            f"Write a concise analyst note for this record: "
            f"features={features}, treatment={treatment}, conversion={conversion}, "
            f"visit={visit}, exposure={exposure}. Do not invent ROI or uplift."
        )
        answer = (
            f"Analyst note: observed treatment={treatment}; conversion={conversion}; "
            f"visit={visit}; exposure={exposure}. Feature values: {features}. "
            "No ROI, incremental impact, or individual uplift can be calculated "
            "from this record alone."
        )

    else:
        prompt = (
            f"What additional analysis is needed for this marketing record? "
            f"features={features}, treatment={treatment}, conversion={conversion}, "
            f"visit={visit}, exposure={exposure}."
        )
        answer = (
            "Calculate treatment and control outcome rates, inspect sample sizes "
            "and randomization, estimate uncertainty, and evaluate relevant "
            "segments. This record alone cannot establish an incremental campaign effect."
        )

    return {
        "messages": [
            {"role": "user", "content": prompt},
            {"role": "assistant", "content": answer}
        ]
    }

examples = []

for _, row in df.iterrows():
    # Multiple task formats per record; not independent experiments.
    for task_type in range(4):
        examples.append(build_example(row, task_type))

rng.shuffle(examples)

print("Source records:", len(df))
print("Generated task examples:", len(examples))
print("Example:\n", json.dumps(examples[0], indent=2))
```

```python
from sklearn.model_selection import train_test_split
from tqdm.auto import tqdm
import numpy as np
import json

groups = np.arange(len(examples)) // 4
unique_groups = np.unique(groups)

train_groups, remaining_groups = train_test_split(
    unique_groups, test_size=0.30, random_state=42
)

valid_groups, test_groups = train_test_split(
    remaining_groups, test_size=0.50, random_state=42
)

train_set = set(train_groups)
valid_set = set(valid_groups)
test_set = set(test_groups)

splits = {
    "train.jsonl": [],
    "valid.jsonl": [],
    "test.jsonl": []
}

# Group examples into splits with a progress bar
for i, item in enumerate(tqdm(examples, desc="Assigning examples")):
    group_id = groups[i]

    if group_id in train_set:
        splits["train.jsonl"].append(item)
    elif group_id in valid_set:
        splits["valid.jsonl"].append(item)
    else:
        splits["test.jsonl"].append(item)

# Write files with progress bars
for filename, items in splits.items():
    path = OUTPUT_DIR / filename

    with open(path, "w", encoding="utf-8") as f:
        for item in tqdm(items, desc=f"Writing {filename}", unit="example"):
            f.write(json.dumps(item, ensure_ascii=False) + "\n")

    print(f"Saved {filename}: {len(items):,} examples")
```

```python
import re
from collections import Counter

def clean_output(text):
    text = re.sub(r"<think>.*?</think>", "", text, flags=re.S)
    text = re.sub(r"=+", "", text)
    text = re.sub(r"Prompt:.*", "", text, flags=re.S)
    text = re.sub(r"Generation:.*", "", text, flags=re.S)
    return " ".join(text.lower().split()).strip()

def word_f1(expected, generated):
    expected_words = clean_output(expected).split()
    generated_words = clean_output(generated).split()

    if not expected_words or not generated_words:
        return 0.0

    common = sum((Counter(expected_words) & Counter(generated_words)).values())

    precision = common / len(generated_words)
    recall = common / len(expected_words)

    return (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0.0
    )

valid = [x for x in evaluation if x["returncode"] == 0]

exact_matches = sum(
    clean_output(x["expected"]) == clean_output(x["generated"])
    for x in valid
)

f1_scores = [
    word_f1(x["expected"], x["generated"])
    for x in valid
]

print("Valid generations:", len(valid))
print(f"Exact match: {exact_matches / max(len(valid), 1):.2%}")
print(f"Average word F1: {sum(f1_scores) / max(len(f1_scores), 1):.2%}")
```

```python
def calculate_lift(treatment_rate, control_rate):
    absolute = treatment_rate - control_rate
    relative = (
        absolute / control_rate * 100
        if control_rate != 0 else None
    )

    return {
        "treatment_rate": treatment_rate,
        "control_rate": control_rate,
        "absolute_lift_percentage_points": absolute * 100,
        "relative_lift_percent": relative
    }

def explain_lift(metrics):
    relative = metrics["relative_lift_percent"]
    relative_text = (
        f"{relative:.2f}%"
        if relative is not None else "undefined because control rate is zero"
    )

    return (
        f"Treatment conversion rate: {metrics['treatment_rate']:.2%}. "
        f"Control conversion rate: {metrics['control_rate']:.2%}. "
        f"Absolute lift: {metrics['absolute_lift_percentage_points']:.2f} "
        f"percentage points. Relative lift: {relative_text}. "
        "This is an observed rate difference; statistical uncertainty and "
        "experiment design should be checked before making a decision."
    )

metrics = calculate_lift(0.08, 0.06)
print(explain_lift(metrics))
```

```python
import subprocess
import json

def causalmark_analyze(treatment_rate, control_rate, campaign_cost=None,
                       incremental_revenue=None):

    # 1. Calculate metrics in Python
    metrics = calculate_lift(treatment_rate, control_rate)

    # 2. Add business metrics only when supplied
    if campaign_cost is not None:
        metrics["campaign_cost"] = campaign_cost

    if incremental_revenue is not None:
        metrics["incremental_revenue"] = incremental_revenue

        if campaign_cost is not None and campaign_cost > 0:
            metrics["incremental_roi_percent"] = (
                (incremental_revenue - campaign_cost) / campaign_cost
            ) * 100

    # 3. Give computed metrics to the fine-tuned model
    prompt = f"""
You are CausalMark, a marketing experiment analyst.

Explain these precomputed metrics to a marketing manager:
{json.dumps(metrics, indent=2)}

Rules:
- Preserve all supplied numeric values.
- Do not invent data or claim statistical significance.
- Explain absolute lift in percentage points.
- Explain relative lift as a percentage.
- If ROI is unavailable, say the supplied data is insufficient to calculate it.
- Recommend checking sample sizes and statistical uncertainty.
""".strip()

    cmd = [
        str(PYTHON), "-m", "mlx_lm", "generate",
        "--model", "mlx-community/Qwen3-8B-4bit",
        "--adapter-path", str(ADAPTER_DIR),
        "--prompt", prompt,
        "--max-tokens", "220",
        "--temp", "0"
    ]

    result = subprocess.run(cmd, capture_output=True, text=True)

    if result.returncode != 0:
        raise RuntimeError(result.stderr)

    return {
        "metrics": metrics,
        "analysis": result.stdout
    }


result = causalmark_analyze(
    treatment_rate=0.08,
    control_rate=0.06
)

print("COMPUTED METRICS:")
print(json.dumps(result["metrics"], indent=2))

print("\nCAUSALMARK ANALYSIS:")
print(result["analysis"])
```

```python
from statsmodels.stats.proportion import confint_proportions_2indep
from scipy.stats import norm
import numpy as np

def analyze_ab_test(treatment_conversions, treatment_n,
                    control_conversions, control_n):

    p_t = treatment_conversions / treatment_n
    p_c = control_conversions / control_n

    difference = p_t - p_c
    relative_lift = difference / p_c * 100 if p_c > 0 else None

    # 95% confidence interval for the difference in proportions
    ci_low, ci_high = confint_proportions_2indep(
        treatment_conversions,
        treatment_n,
        control_conversions,
        control_n,
        method="newcomb",
        alpha=0.05
    )

    # Two-sided pooled two-proportion z-test
    pooled = (
        treatment_conversions + control_conversions
    ) / (treatment_n + control_n)

    standard_error = np.sqrt(
        pooled * (1 - pooled) * (1 / treatment_n + 1 / control_n)
    )

    z_score = difference / standard_error if standard_error > 0 else 0
    p_value = 2 * norm.sf(abs(z_score))

    return {
        "treatment_rate": p_t,
        "control_rate": p_c,
        "absolute_lift_pp": difference * 100,
        "relative_lift_percent": relative_lift,
        "ci_95_low_pp": ci_low * 100,
        "ci_95_high_pp": ci_high * 100,
        "z_score": z_score,
        "p_value": p_value,
        "statistically_significant_at_5_percent": p_value < 0.05
    }


# Example: 800 conversions out of 10,000 vs 600 out of 10,000
ab_result = analyze_ab_test(
    treatment_conversions=800,
    treatment_n=10_000,
    control_conversions=600,
    control_n=10_000
)

print(json.dumps(ab_result, indent=2, default=lambda x: x.item()))
```

```python
import json
import numpy as np
import subprocess

# Convert NumPy values into standard Python values
ab_result = {
    key: value.item() if isinstance(value, np.generic) else value
    for key, value in ab_result.items()
}

def explain_ab_test(ab_result):
    prompt = f"""
You are CausalMark, a marketing experiment analyst.

Interpret these computed A/B test results:
{json.dumps(ab_result, indent=2)}

Provide:
1. A concise results summary.
2. What the 95% confidence interval means.
3. Whether the result is statistically significant at the 5% level.
4. Limitations and a practical next step.

Rules:
- Use the supplied numbers exactly.
- Do not invent revenue, profit, or ROI.
- Distinguish statistical significance from business significance.
- Keep the answer concise and practical.
""".strip()

    cmd = [
        str(PYTHON), "-m", "mlx_lm", "generate",
        "--model", "mlx-community/Qwen3-8B-4bit",
        "--adapter-path", str(ADAPTER_DIR),
        "--prompt", prompt,
        "--max-tokens", "250",
        "--temp", "0"
    ]

    result = subprocess.run(
        cmd,
        capture_output=True,
        text=True
    )

    if result.returncode != 0:
        raise RuntimeError(result.stderr)

    return result.stdout

print(explain_ab_test(ab_result))
```

```python
import json
import numpy as np

def run_causalmark_experiment(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users
):
    # 1. Validate inputs
    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("User counts must be greater than zero.")

    if not (0 <= treatment_conversions <= treatment_users):
        raise ValueError("Invalid treatment conversion count.")

    if not (0 <= control_conversions <= control_users):
        raise ValueError("Invalid control conversion count.")

    # 2. Calculate metrics
    result = analyze_ab_test(
        treatment_conversions,
        treatment_users,
        control_conversions,
        control_users
    )

    # 3. Convert NumPy values for JSON compatibility
    result = {
        key: value.item() if isinstance(value, np.generic) else value
        for key, value in result.items()
    }

    # 4. Generate AI interpretation
    explanation = explain_ab_test(result)

    return {
        "metrics": result,
        "ai_analysis": explanation
    }


# Example experiment
experiment = run_causalmark_experiment(
    treatment_conversions=800,
    treatment_users=10000,
    control_conversions=600,
    control_users=10000
)

print("=== CAUSALMARK METRICS ===")
print(json.dumps(experiment["metrics"], indent=2))

print("\n=== AI ANALYSIS ===")
print(experiment["ai_analysis"])
```

```python
import numpy as np
import pandas as pd

# Illustrative assumptions — replace with actual campaign costs
campaign_costs = {
    "Campaign A": 5.0,
    "Campaign B": 4.0,
    "Campaign C": 3.0
}

TOTAL_BUDGET = 100_000

budget_df = ranking_df.copy()

budget_df["cost_per_user"] = budget_df["experiment"].map(campaign_costs)

# Only consider campaigns that passed the adjusted significance check
budget_df = budget_df[
    budget_df["eligible_for_scaling_review"]
].copy()

if budget_df.empty:
    print("No campaigns currently qualify for budget allocation.")
else:
    # Estimated incremental value per unit of campaign spend
    budget_df["value_per_spend"] = (
        budget_df["lift_pp"] / 100
        * budget_df["value_per_conversion"]
        / budget_df["cost_per_user"]
    )

    budget_df = budget_df.sort_values(
        "value_per_spend", ascending=False
    ).reset_index(drop=True)

    # Allocate budget greedily, subject to the remaining budget
    remaining_budget = TOTAL_BUDGET
    allocations = []

    for _, row in budget_df.iterrows():
        spend = min(
            remaining_budget,
            TOTAL_BUDGET / len(budget_df)
        )

        allocations.append(spend)
        remaining_budget -= spend

    budget_df["allocated_budget"] = allocations
    budget_df["estimated_reach"] = (
        budget_df["allocated_budget"] / budget_df["cost_per_user"]
    )

    budget_df["estimated_incremental_conversions"] = (
        budget_df["estimated_reach"] * budget_df["lift_pp"] / 100
    )

    budget_df["estimated_incremental_value"] = (
        budget_df["estimated_incremental_conversions"]
        * budget_df["value_per_conversion"]
    )

    display(
        budget_df[
            [
                "experiment",
                "value_per_spend",
                "allocated_budget",
                "estimated_reach",
                "estimated_incremental_conversions",
                "estimated_incremental_value"
            ]
        ]
    )

    print(f"Total allocated: {budget_df['allocated_budget'].sum():,.2f}")
    print(f"Budget remaining: {remaining_budget:,.2f}")
```

```python
from scipy.optimize import linprog
import numpy as np
import pandas as pd

# Illustrative assumptions: replace with actual campaign data
campaign_data = pd.DataFrame([
    {"experiment": "Campaign A", "lift": 0.02, "value_per_conversion": 200,
     "cost_per_user": 5, "max_users": 10000},
    {"experiment": "Campaign B", "lift": 0.01, "value_per_conversion": 200,
     "cost_per_user": 4, "max_users": 9000},
    {"experiment": "Campaign C", "lift": 0.01, "value_per_conversion": 200,
     "cost_per_user": 3, "max_users": 6000},
])

TOTAL_BUDGET = 100_000

# Expected incremental value per additional user
campaign_data["value_per_user"] = (
    campaign_data["lift"] * campaign_data["value_per_conversion"]
)

# Net incremental value per user after campaign cost
campaign_data["net_value_per_user"] = (
    campaign_data["value_per_user"] - campaign_data["cost_per_user"]
)

# Maximize net value by minimizing its negative
objective = -campaign_data["net_value_per_user"].to_numpy()

# Budget constraint: sum(cost_per_user * users) <= TOTAL_BUDGET
A_ub = [campaign_data["cost_per_user"].to_numpy()]
b_ub = [TOTAL_BUDGET]

# Each campaign must stay within its capacity
bounds = [
    (0, int(max_users))
    for max_users in campaign_data["max_users"]
]

result = linprog(
    c=objective,
    A_ub=A_ub,
    b_ub=b_ub,
    bounds=bounds,
    method="highs"
)

if not result.success:
    raise RuntimeError(result.message)

campaign_data["recommended_users"] = np.floor(result.x).astype(int)
campaign_data["recommended_spend"] = (
    campaign_data["recommended_users"] * campaign_data["cost_per_user"]
)
campaign_data["expected_incremental_value"] = (
    campaign_data["recommended_users"] * campaign_data["value_per_user"]
)
campaign_data["expected_net_value"] = (
    campaign_data["recommended_users"] * campaign_data["net_value_per_user"]
)

display(campaign_data[[
    "experiment",
    "recommended_users",
    "recommended_spend",
    "expected_incremental_value",
    "expected_net_value"
]])

print(f"Total spend: {campaign_data['recommended_spend'].sum():,.2f}")
print(f"Budget remaining: {TOTAL_BUDGET - campaign_data['recommended_spend'].sum():,.2f}")
print(f"Expected net value: {campaign_data['expected_net_value'].sum():,.2f}")
```

```python
total_spend = campaign_data["recommended_spend"].sum()

assert total_spend <= TOTAL_BUDGET
assert (
    campaign_data["recommended_users"]
    <= campaign_data["max_users"]
).all()
assert (campaign_data["recommended_users"] >= 0).all()

print("Budget constraint: PASS")
print("Campaign capacity constraints: PASS")
print("Non-negative allocations: PASS")
```

```python
import pandas as pd
import numpy as np
from scipy.optimize import linprog

TOTAL_BUDGET = 100_000

def optimize_budget(lift_multiplier):
    df = campaign_data.copy()

    # Adjust the observed lift for this scenario
    df["scenario_lift"] = df["lift"] * lift_multiplier

    df["scenario_net_value"] = (
        df["scenario_lift"] * df["value_per_conversion"]
        - df["cost_per_user"]
    )

    result = linprog(
        c=-df["scenario_net_value"].to_numpy(),
        A_ub=[df["cost_per_user"].to_numpy()],
        b_ub=[TOTAL_BUDGET],
        bounds=[
            (0, int(capacity))
            for capacity in df["max_users"]
        ],
        method="highs"
    )

    if not result.success:
        raise RuntimeError(result.message)

    df["allocated_users"] = np.floor(result.x).astype(int)
    df["allocated_spend"] = (
        df["allocated_users"] * df["cost_per_user"]
    )
    df["expected_net_value"] = (
        df["allocated_users"] * df["scenario_net_value"]
    )

    return df


scenarios = {
    "Pessimistic (-30% lift)": 0.70,
    "Expected lift": 1.00,
    "Optimistic (+30% lift)": 1.30
}

summary = []

for scenario, multiplier in scenarios.items():
    allocation = optimize_budget(multiplier)

    summary.append({
        "scenario": scenario,
        "budget_used": allocation["allocated_spend"].sum(),
        "expected_net_value": allocation["expected_net_value"].sum(),
        "campaigns_funded": int((allocation["allocated_users"] > 0).sum())
    })

sensitivity_df = pd.DataFrame(summary)
display(sensitivity_df)
```

```python
def evaluate_experiment(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    alpha=0.05
):
    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("User counts must be positive.")

    if not 0 <= treatment_conversions <= treatment_users:
        raise ValueError("Invalid treatment conversion count.")

    if not 0 <= control_conversions <= control_users:
        raise ValueError("Invalid control conversion count.")

    result = analyze_ab_test(
        treatment_conversions,
        treatment_users,
        control_conversions,
        control_users
    )

    result = {
        key: value.item() if hasattr(value, "item") else value
        for key, value in result.items()
    }

    result["decision"] = (
        "Positive lift; statistically significant"
        if result["absolute_lift_pp"] > 0
        and result["p_value"] < alpha
        else "Insufficient evidence of significant positive lift"
    )

    return result


example_result = evaluate_experiment(
    treatment_conversions=800,
    treatment_users=10000,
    control_conversions=600,
    control_users=10000
)

print(json.dumps(example_result, indent=2))
```

```python
import numpy as np
from scipy.optimize import linprog

TOTAL_BUDGET = 100_000

df = campaign_data.copy()

objective = -df["net_value_per_user"].to_numpy()

optimizer_result = linprog(
    c=objective,
    A_ub=[df["cost_per_user"].to_numpy()],
    b_ub=[TOTAL_BUDGET],
    bounds=[
        (0, int(capacity))
        for capacity in df["max_users"]
    ],
    method="highs"
)

if not optimizer_result.success:
    raise RuntimeError(optimizer_result.message)

df["recommended_users"] = np.floor(
    optimizer_result.x
).astype(int)

df["recommended_spend"] = (
    df["recommended_users"] * df["cost_per_user"]
)

assert df["recommended_spend"].sum() <= TOTAL_BUDGET
assert (df["recommended_users"] >= 0).all()
assert (
    df["recommended_users"] <= df["max_users"]
).all()

display(df[[
    "experiment",
    "recommended_users",
    "recommended_spend"
]])

print("Budget optimizer checks passed.")
print(f"Budget allocated: {df['recommended_spend'].sum():,.2f}")
print(
    f"Budget remaining: "
    f"{TOTAL_BUDGET - df['recommended_spend'].sum():,.2f}"
)
```

```python
import json
import re
import pandas as pd
from pathlib import Path

data_dir = PROJECT_DIR / "data" / "marketing_instructions"
results_path = data_dir / "evaluation_100.jsonl"

with open(results_path, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

def normalize(text):
    text = re.sub(r"<think>.*?</think>", "", str(text), flags=re.S)
    text = re.sub(r"\s+", " ", text)
    return text.strip().lower()

evaluation = pd.DataFrame([
    {
        "prompt": r.get("prompt", ""),
        "expected": r.get("expected", ""),
        "generated": r.get("generated", ""),
        "returncode": r.get("returncode", None),
        "exact_match": normalize(r.get("expected", "")) == normalize(r.get("generated", "")),
        "output_length": len(str(r.get("generated", "")).strip())
    }
    for r in records
])

print("Evaluated examples:", len(evaluation))
print("Exact matches:", int(evaluation["exact_match"].sum()))
print("Exact-match rate:", round(evaluation["exact_match"].mean(), 4))
print("Empty outputs:", int((evaluation["output_length"] == 0).sum()))
print("Failed commands:", int((evaluation["returncode"] != 0).sum()))

display(evaluation[["prompt", "expected", "generated", "exact_match"]].head(5))
```

```python
import re
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def clean_text(text):
    text = re.sub(r"<think>.*?</think>", "", str(text), flags=re.S)
    text = re.sub(r"=+", "", text)
    text = re.sub(r"^(prompt|generation):", "", text, flags=re.I)
    return " ".join(text.lower().split()).strip()

evaluation["clean_expected"] = evaluation["expected"].apply(clean_text)
evaluation["clean_generated"] = evaluation["generated"].apply(clean_text)

vectorizer = TfidfVectorizer(ngram_range=(1, 2), stop_words="english")

all_text = (
    evaluation["clean_expected"].tolist()
    + evaluation["clean_generated"].tolist()
)

matrix = vectorizer.fit_transform(all_text)

n = len(evaluation)
similarities = cosine_similarity(
    matrix[:n],
    matrix[n:]
).diagonal()

evaluation["tfidf_similarity"] = similarities

print("Mean TF-IDF similarity:", round(float(np.mean(similarities)), 4))
print("Median similarity:", round(float(np.median(similarities)), 4))
print("Similarity >= 0.5:", int((similarities >= 0.5).sum()), "/", n)

display(
    evaluation[
        ["clean_expected", "clean_generated", "tfidf_similarity"]
    ].sort_values("tfidf_similarity", ascending=False).head(10)
)
```

```python
import re
import pandas as pd

def extract_binary_value(text, field):
    patterns = [
        rf"\b{field}\s*[:=]\s*([01])\b",
        rf"\b{field}\s+(?:is\s+)?([01])\b"
    ]

    for pattern in patterns:
        match = re.search(pattern, str(text), re.I)
        if match:
            return int(match.group(1))

    return None

checks = []

for _, row in evaluation.iterrows():
    expected = row["clean_expected"]
    generated = row["clean_generated"]

    record = {"prompt": row["prompt"]}

    for field in ["treatment", "conversion", "visit", "exposure"]:
        expected_value = extract_binary_value(expected, field)
        generated_value = extract_binary_value(generated, field)

        record[f"{field}_expected"] = expected_value
        record[f"{field}_generated"] = generated_value
        record[f"{field}_correct"] = (
            expected_value == generated_value
            if expected_value is not None and generated_value is not None
            else None
        )

    checks.append(record)

factual_df = pd.DataFrame(checks)

for field in ["treatment", "conversion", "visit", "exposure"]:
    column = f"{field}_correct"
    valid = factual_df[column].dropna()

    if len(valid):
        print(
            f"{field}: {int(valid.sum())}/{len(valid)} correct "
            f"({valid.mean():.1%})"
        )
    else:
        print(f"{field}: no extractable values")

display(factual_df.head(10))
```

```python
import json
import re
import pandas as pd
from pathlib import Path

results_path = DATA_DIR / "heldout_benchmark_results.jsonl"

with open(results_path, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

def clean_answer(text):
    text = re.sub(r"<think>.*?</think>", "", str(text), flags=re.S)
    text = re.sub(r"=+", "", text)
    text = re.sub(r"\n*Prompt:\s*\d+ tokens.*$", "", text, flags=re.S)
    text = re.sub(r"\n*Generation:\s*\d+ tokens.*$", "", text, flags=re.S)
    return text.strip()

def has_number(text, expected, tolerance=0.02):
    values = re.findall(r"[-+]?\d*\.?\d+", text.replace(",", ""))
    return any(abs(float(v) - expected) <= tolerance for v in values)

checks = []

for i, record in enumerate(records):
    answer = clean_answer(record["generated"])
    expected = record["expected"]

    row = {
        "question": i + 1,
        "answer_length": len(answer),
        "repetition_ratio": (
            len(answer.split()) / max(1, len(set(answer.split())))
        ),
        "answer": answer[:180]
    }

    if "absolute_lift_pp" in expected:
        row["absolute_lift_correct"] = has_number(
            answer, expected["absolute_lift_pp"]
        )

    if "relative_lift_pct" in expected:
        row["relative_lift_correct"] = has_number(
            answer, expected["relative_lift_pct"], tolerance=0.15
        )

    if "cost_per_incremental_purchase" in expected:
        row["cost_per_purchase_correct"] = has_number(
            answer, expected["cost_per_incremental_purchase"]
        )

    if "significant" in expected:
        row["significance_correct"] = (
            "not statistically significant" in answer.lower()
            or "should not be declared statistically significant" in answer.lower()
        )

    checks.append(row)

score_df = pd.DataFrame(checks)
display(score_df.drop(columns=["answer"]))
display(score_df[["question", "answer"]])

print("\nNumerical checks:")
for column in [
    "absolute_lift_correct",
    "relative_lift_correct",
    "cost_per_purchase_correct",
    "significance_correct"
]:
    if column in score_df:
        valid = score_df[column].dropna()
        print(f"{column}: {int(valid.sum())}/{len(valid)}")
```

```python
import json
import re
import pandas as pd
from pathlib import Path

DATA_DIR = PROJECT_DIR / "data" / "marketing_instructions"
RESULTS_PATH = DATA_DIR / "heldout_benchmark_controlled.jsonl"

with open(RESULTS_PATH, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

def extract_percentages(text):
    return [
        float(x)
        for x in re.findall(r"(?<![\w.])(-?\d+(?:\.\d+)?)\s*%", text)
    ]

def extract_lift_values(text):
    patterns = [
        r"absolute lift[^.\n:]*[:=]\s*(-?\d+(?:\.\d+)?)\s*(?:percentage points|pp)",
        r"relative lift[^.\n:]*[:=]\s*(-?\d+(?:\.\d+)?)\s*%"
    ]

    values = {}
    for name, pattern in zip(["absolute_lift_pp", "relative_lift_pct"], patterns):
        match = re.search(pattern, text, re.I)
        values[name] = float(match.group(1)) if match else None

    return values

rows = []

for i, record in enumerate(records, start=1):
    answer = record["generated"]
    expected = record["expected"]

    row = {
        "question": i,
        "generation_succeeded": record["returncode"] == 0,
        "answer_length": len(answer),
        "repeated_lines": len(answer.splitlines()) - len(set(answer.splitlines()))
    }

    lift_values = extract_lift_values(answer)

    for metric in ["absolute_lift_pp", "relative_lift_pct"]:
        if metric in expected:
            row[metric + "_expected"] = expected[metric]
            row[metric + "_predicted"] = lift_values[metric]
            row[metric + "_correct"] = (
                lift_values[metric] is not None
                and abs(lift_values[metric] - expected[metric]) < 0.15
            )

    if "cost_per_incremental_purchase" in expected:
        match = re.search(
            r"(?:cost per incremental purchase|per incremental purchase)"
            r"[^0-9$]*\$?\s*(\d+(?:\.\d+)?)",
            answer,
            re.I
        )
        predicted = float(match.group(1)) if match else None
        row["cost_expected"] = expected["cost_per_incremental_purchase"]
        row["cost_predicted"] = predicted
        row["cost_correct"] = predicted is not None and abs(predicted - 15) < 0.01

    rows.append(row)

score_df = pd.DataFrame(rows)
display(score_df)

print("\nSummary:")
for column in [
    "absolute_lift_pp_correct",
    "relative_lift_pct_correct",
    "cost_correct"
]:
    if column in score_df:
        valid = score_df[column].dropna()
        if len(valid):
            print(f"{column}: {int(valid.sum())}/{len(valid)}")
```

```python
import pandas as pd

cases = [
    {
        "question": 1,
        "treatment_conversions": 450,
        "treatment_users": 5000,
        "control_conversions": 350,
        "control_users": 5000
    },
    {
        "question": 2,
        "treatment_conversions": 120,
        "treatment_users": 2000,
        "control_conversions": 100,
        "control_users": 2000
    },
    {
        "question": 4,
        "treatment_conversions": 800,
        "treatment_users": 10000,
        "control_conversions": 700,
        "control_users": 10000
    },
    {
        "question": 10,
        "treatment_rate": 0.08,
        "control_rate": 0.06
    }
]

verified = []

for case in cases:
    if "treatment_conversions" in case:
        treatment_rate = case["treatment_conversions"] / case["treatment_users"]
        control_rate = case["control_conversions"] / case["control_users"]
    else:
        treatment_rate = case["treatment_rate"]
        control_rate = case["control_rate"]

    absolute_lift_pp = (treatment_rate - control_rate) * 100
    relative_lift_pct = (
        (treatment_rate - control_rate) / control_rate * 100
        if control_rate > 0 else None
    )

    verified.append({
        "question": case["question"],
        "treatment_rate_pct": treatment_rate * 100,
        "control_rate_pct": control_rate * 100,
        "absolute_lift_pp": absolute_lift_pp,
        "relative_lift_pct": relative_lift_pct
    })

verified_df = pd.DataFrame(verified)
display(verified_df.round(4))

cost_per_purchase = 3600 / 240
print(f"Question 3 cost per incremental purchase: ${cost_per_purchase:.2f}")
print("Question 5 significant at alpha=0.05:", 0.18 < 0.05)
```

```python
import json
import re
import pandas as pd
from pathlib import Path

RESULTS_PATH = DATA_DIR / "heldout_benchmark_controlled.jsonl"

with open(RESULTS_PATH, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

def answer_text(text):
    text = re.sub(r"<think>.*?</think>", "", str(text), flags=re.S)
    text = re.sub(r"=+", "", text)
    text = re.sub(r"\n*(Prompt|Generation):.*$", "", text, flags=re.S)
    return text.strip()

def labeled_value(text, labels):
    for label in labels:
        pattern = rf"{label}\s*[:=]\s*(-?\d+(?:\.\d+)?)\s*(%|percentage points|pp)?"
        match = re.search(pattern, text, re.I)
        if match:
            return float(match.group(1)), (match.group(2) or "").lower()
    return None, None

checks = []

for i, record in enumerate(records, start=1):
    answer = answer_text(record["generated"])
    expected = record["expected"]
    row = {"question": i, "answer": answer[:250]}

    if "absolute_lift_pp" in expected:
        value, unit = labeled_value(
            answer, [r"absolute lift", r"absolute increase"]
        )
        row["absolute_lift_expected"] = expected["absolute_lift_pp"]
        row["absolute_lift_predicted"] = value
        row["absolute_lift_correct"] = (
            value is not None
            and unit in ["percentage points", "pp", "%"]
            and abs(value - expected["absolute_lift_pp"]) < 0.15
        )

    if "relative_lift_pct" in expected:
        value, unit = labeled_value(
            answer, [r"relative lift", r"relative increase"]
        )
        row["relative_lift_expected"] = expected["relative_lift_pct"]
        row["relative_lift_predicted"] = value
        row["relative_lift_correct"] = (
            value is not None
            and unit == "%"
            and abs(value - expected["relative_lift_pct"]) < 0.15
        )

    if "cost_per_incremental_purchase" in expected:
        match = re.search(
            r"cost per incremental purchase\s*(?:is|=|:)?\s*\$?(\d+(?:\.\d+)?)",
            answer, re.I
        )
        value = float(match.group(1)) if match else None
        row["cost_expected"] = expected["cost_per_incremental_purchase"]
        row["cost_predicted"] = value
        row["cost_correct"] = value is not None and abs(value - 15) < 0.01

    checks.append(row)

checks_df = pd.DataFrame(checks)
display(checks_df)

print("\nScored metrics:")
for column in [
    "absolute_lift_correct",
    "relative_lift_correct",
    "cost_correct"
]:
    if column in checks_df:
        valid = checks_df[column].dropna()
        print(f"{column}: {int(valid.sum())}/{len(valid)} scored as correct")
```

```python
import json
import re
from pathlib import Path

DATA_DIR = PROJECT_DIR / "data" / "marketing_instructions"
INPUT_PATH = DATA_DIR / "numerical_marketing_examples.jsonl"

with open(INPUT_PATH, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

def extract_number(text, pattern):
    match = re.search(pattern, text, re.I)
    return float(match.group(1)) if match else None

failures = []

for i, record in enumerate(records, start=1):
    prompt = record["messages"][0]["content"]
    answer = record["messages"][1]["content"]

    values = {
        "treatment": re.search(
            r"(\d+) conversions among (\d+) treatment users", prompt
        ),
        "control": re.search(
            r"(\d+) conversions among (\d+) control users", prompt
        )
    }

    if not all(values.values()):
        failures.append((i, "Could not parse prompt"))
        continue

    tc, tn = map(int, values["treatment"].groups())
    cc, cn = map(int, values["control"].groups())

    tr = tc / tn * 100
    cr = cc / cn * 100
    lift = tr - cr
    relative = lift / cr * 100

    answer_values = [
        extract_number(answer, r"Treatment conversion rate.*?=\s*([\d.]+)%"),
        extract_number(answer, r"Control conversion rate.*?=\s*([\d.]+)%"),
        extract_number(answer, r"Absolute lift.*?=\s*([-+]?[\d.]+)\s+percentage points"),
        extract_number(answer, r"Relative lift.*?=\s*([-+]?[\d.]+)%")
    ]

    expected = [tr, cr, lift, relative]

    if any(v is None for v in answer_values):
        failures.append((i, "Missing numerical value"))
    elif any(abs(a - e) > 0.0002 for a, e in zip(answer_values, expected)):
        failures.append((i, "Numerical mismatch"))

print("Total examples:", len(records))
print("Valid examples:", len(records) - len(failures))
print("Failed examples:", len(failures))

if failures:
    print("First failures:", failures[:10])
else:
    print("All numerical examples passed validation.")
```

```python
import json
import re
from pathlib import Path

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
RESULTS_PATH = PROJECT_DIR / "data" / "marketing_instructions" / "numerical_stress_test_results.jsonl"

with open(RESULTS_PATH, "r", encoding="utf-8") as f:
    results = [json.loads(line) for line in f if line.strip()]

def extract_metric(text, metric):
    text = re.sub(r"<think>.*?</think>", "", text, flags=re.S)
    text = text.replace(",", "")
    patterns = {
        "treatment_rate_pct": r"(?:treatment conversion rate|treatment rate)\s*[:=]?\s*([+-]?\d+(?:\.\d+)?)\s*%",
        "control_rate_pct": r"(?:control conversion rate|control rate)\s*[:=]?\s*([+-]?\d+(?:\.\d+)?)\s*%",
        "absolute_lift_pp": r"(?:absolute lift|absolute difference)\s*[:=]?\s*([+-]?\d+(?:\.\d+)?)\s*(?:percentage points|pp)\b",
        "relative_lift_pct": r"(?:relative lift)\s*[:=]?\s*([+-]?\d+(?:\.\d+)?)\s*%"
    }
    match = re.search(patterns[metric], text, re.I)
    return float(match.group(1)) if match else None

metrics = [
    "treatment_rate_pct",
    "control_rate_pct",
    "absolute_lift_pp",
    "relative_lift_pct"
]

totals = {m: 0 for m in metrics}
correct = {m: 0 for m in metrics}
review = []

for i, item in enumerate(results, 1):
    answer = item.get("generated", "")
    expected = item["expected"]
    case_passed = True

    for metric in metrics:
        predicted = extract_metric(answer, metric)
        target = expected[metric]

        if predicted is not None:
            totals[metric] += 1
            if abs(predicted - target) <= 0.02:
                correct[metric] += 1
            else:
                case_passed = False
        else:
            case_passed = False

    if not case_passed:
        review.append({
            "case": i,
            "expected": expected,
            "generated": answer[:700]
        })

print("NUMERICAL STRESS TEST")
print("=" * 55)

for metric in metrics:
    attempted = totals[metric]
    passed = correct[metric]
    accuracy = 100 * passed / attempted if attempted else 0
    print(f"{metric}: {passed}/{attempted} correct among parsed answers ({accuracy:.1f}%)")

print("\nCases needing review:", len(review))
print("Overall exact-case passes:", len(results) - len(review), "/", len(results))

for item in review[:10]:
    print(f"\nCASE {item['case']}")
    print("Expected:", item["expected"])
    print("Generated:", item["generated"])
```

```python
import json
import re
from collections import Counter
from pathlib import Path

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
RESULTS_PATH = PROJECT_DIR / "data" / "marketing_instructions" / "numerical_stress_test_results.jsonl"

with open(RESULTS_PATH, "r", encoding="utf-8") as f:
    results = [json.loads(line) for line in f if line.strip()]

metrics = {
    "treatment_rate_pct": r"(?:treatment conversion rate|treatment rate)\s*[:=]?\s*([\d.]+)\s*%",
    "control_rate_pct": r"(?:control conversion rate|control rate)\s*[:=]?\s*([\d.]+)\s*%",
    "absolute_lift_pp": r"(?:absolute lift|absolute difference)\s*[:=]?\s*([\d.]+)\s*(?:percentage points|pp)\b",
    "relative_lift_pct": r"relative lift\s*[:=]?\s*([\d.]+)\s*%"
}

summary = Counter()
failures = []

for index, item in enumerate(results, 1):
    answer = re.sub(r"<think>.*?</think>", "", item.get("generated", ""), flags=re.S)
    expected = item["expected"]
    failed_metrics = []

    for metric, pattern in metrics.items():
        match = re.search(pattern, answer, re.I)
        if not match:
            summary[f"{metric}: not parsed"] += 1
            failed_metrics.append(f"{metric}: not parsed")
            continue

        predicted = float(match.group(1))
        target = expected[metric]

        if abs(predicted - target) <= 0.02:
            summary[f"{metric}: correct"] += 1
        else:
            summary[f"{metric}: incorrect"] += 1
            failed_metrics.append(
                f"{metric}: predicted {predicted}, expected {target}"
            )

    if failed_metrics:
        failures.append({
            "case": index,
            "question": item["question"],
            "issues": failed_metrics,
            "answer": answer[:500]
        })

print("ERROR SUMMARY")
print("=" * 60)

for category, count in sorted(summary.items()):
    print(f"{category}: {count}")

print(f"\nCases with at least one issue: {len(failures)}/{len(results)}")

for failure in failures[:10]:
    print(f"\nCase {failure['case']}")
    print("Issues:", "; ".join(failure["issues"]))
    print("Answer:", failure["answer"])
```

```python
def calculate_lift(treatment_conversions, treatment_users,
                   control_conversions, control_users):

    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("Group sizes must be positive.")

    if not 0 <= treatment_conversions <= treatment_users:
        raise ValueError("Invalid treatment conversion count.")

    if not 0 <= control_conversions <= control_users:
        raise ValueError("Invalid control conversion count.")

    treatment_rate = treatment_conversions / treatment_users
    control_rate = control_conversions / control_users

    absolute_lift_pp = (treatment_rate - control_rate) * 100

    relative_lift_pct = (
        (treatment_rate - control_rate) / control_rate * 100
        if control_rate > 0 else None
    )

    return {
        "treatment_rate_pct": round(treatment_rate * 100, 4),
        "control_rate_pct": round(control_rate * 100, 4),
        "absolute_lift_pp": round(absolute_lift_pp, 4),
        "relative_lift_pct": (
            round(relative_lift_pct, 4)
            if relative_lift_pct is not None else None
        )
    }


example = calculate_lift(800, 10000, 700, 10000)

print("Calculated results:")
for metric, value in example.items():
    print(f"{metric}: {value}")
```

```python
def calculate_campaign_economics(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    campaign_cost,
    revenue_per_incremental_conversion
):
    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("Group sizes must be positive.")

    if not 0 <= treatment_conversions <= treatment_users:
        raise ValueError("Invalid treatment conversions.")

    if not 0 <= control_conversions <= control_users:
        raise ValueError("Invalid control conversions.")

    if campaign_cost < 0 or revenue_per_incremental_conversion < 0:
        raise ValueError("Cost and revenue must be non-negative.")

    treatment_rate = treatment_conversions / treatment_users
    control_rate = control_conversions / control_users

    incremental_rate = treatment_rate - control_rate
    incremental_conversions = incremental_rate * treatment_users

    cost_per_incremental = (
        campaign_cost / incremental_conversions
        if incremental_conversions > 0 else None
    )

    incremental_revenue = (
        incremental_conversions * revenue_per_incremental_conversion
    )

    incremental_profit = incremental_revenue - campaign_cost

    return {
        "incremental_conversion_rate": round(incremental_rate, 6),
        "estimated_incremental_conversions": round(incremental_conversions, 2),
        "campaign_cost": round(campaign_cost, 2),
        "cost_per_incremental_conversion": (
            round(cost_per_incremental, 2)
            if cost_per_incremental is not None else None
        ),
        "estimated_incremental_revenue": round(incremental_revenue, 2),
        "estimated_incremental_profit": round(incremental_profit, 2)
    }


example = calculate_campaign_economics(
    treatment_conversions=450,
    treatment_users=5000,
    control_conversions=350,
    control_users=5000,
    campaign_cost=3600,
    revenue_per_incremental_conversion=25
)

for metric, value in example.items():
    print(f"{metric}: {value}")
```

```python
from math import sqrt
from statistics import NormalDist

def test_conversion_significance(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    alpha=0.05
):
    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("Group sizes must be positive.")

    if not 0 <= treatment_conversions <= treatment_users:
        raise ValueError("Invalid treatment conversions.")

    if not 0 <= control_conversions <= control_users:
        raise ValueError("Invalid control conversions.")

    if not 0 < alpha < 1:
        raise ValueError("Alpha must be between 0 and 1.")

    p_t = treatment_conversions / treatment_users
    p_c = control_conversions / control_users
    pooled = (
        treatment_conversions + control_conversions
    ) / (treatment_users + control_users)

    standard_error = sqrt(
        pooled * (1 - pooled) *
        (1 / treatment_users + 1 / control_users)
    )

    z_score = (p_t - p_c) / standard_error if standard_error else 0
    p_value = 2 * (1 - NormalDist().cdf(abs(z_score)))

    return {
        "treatment_rate_pct": round(p_t * 100, 4),
        "control_rate_pct": round(p_c * 100, 4),
        "absolute_lift_pp": round((p_t - p_c) * 100, 4),
        "z_score": round(z_score, 4),
        "p_value": round(p_value, 6),
        "alpha": alpha,
        "statistically_significant": p_value < alpha
    }


example = test_conversion_significance(
    treatment_conversions=520,
    treatment_users=10000,
    control_conversions=500,
    control_users=10000
)

for metric, value in example.items():
    print(f"{metric}: {value}")
```

```python
def analyze_experiment(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    campaign_cost=None,
    revenue_per_incremental_conversion=None,
    alpha=0.05
):
    lift = calculate_lift(
        treatment_conversions,
        treatment_users,
        control_conversions,
        control_users
    )

    significance = test_conversion_significance(
        treatment_conversions,
        treatment_users,
        control_conversions,
        control_users,
        alpha
    )

    report = {
        "conversion_analysis": lift,
        "significance_analysis": significance
    }

    if campaign_cost is not None and revenue_per_incremental_conversion is not None:
        report["campaign_economics"] = calculate_campaign_economics(
            treatment_conversions,
            treatment_users,
            control_conversions,
            control_users,
            campaign_cost,
            revenue_per_incremental_conversion
        )

    return report


report = analyze_experiment(
    treatment_conversions=450,
    treatment_users=5000,
    control_conversions=350,
    control_users=5000,
    campaign_cost=3600,
    revenue_per_incremental_conversion=25
)

import json
print(json.dumps(report, indent=2))
```

```python
def generate_experiment_report(report):
    conversion = report["conversion_analysis"]
    significance = report["significance_analysis"]

    lift = conversion["absolute_lift_pp"]
    relative_lift = conversion["relative_lift_pct"]
    p_value = significance["p_value"]
    significant = significance["statistically_significant"]

    if lift <= 0:
        recommendation = "Do not scale based on conversion lift; investigate the negative or zero effect."
    elif not significant:
        recommendation = "Gather more evidence before scaling; the observed lift is not statistically significant."
    else:
        recommendation = "The experiment shows statistically significant positive conversion lift. Evaluate profitability before scaling."

    economics = report.get("campaign_economics")

    if economics:
        profit = economics["estimated_incremental_profit"]

        if profit <= 0:
            recommendation += " Current estimated incremental profit is non-positive; revise campaign economics."
        else:
            recommendation += " Estimated incremental profit is positive under the supplied assumptions."

    lines = [
        "CAUSALMARK EXPERIMENT REPORT",
        "=" * 40,
        f"Treatment conversion: {conversion['treatment_rate_pct']}%",
        f"Control conversion: {conversion['control_rate_pct']}%",
        f"Absolute lift: {lift} percentage points",
        f"Relative lift: {relative_lift}%",
        f"P-value: {p_value}",
        f"Significant at alpha={significance['alpha']}: {significant}",
    ]

    if economics:
        lines.extend([
            f"Estimated incremental conversions: {economics['estimated_incremental_conversions']}",
            f"Campaign cost: {economics['campaign_cost']}",
            f"Estimated incremental revenue: {economics['estimated_incremental_revenue']}",
            f"Estimated incremental profit: {economics['estimated_incremental_profit']}",
        ])

    lines.extend([
        "",
        "RECOMMENDATION",
        recommendation
    ])

    return "\n".join(lines)


print(generate_experiment_report(report))
```

```python
from math import sqrt
from statistics import NormalDist

def calculate_lift_confidence_interval(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    confidence=0.95
):
    if treatment_users <= 0 or control_users <= 0:
        raise ValueError("Group sizes must be positive.")

    if not 0 <= treatment_conversions <= treatment_users:
        raise ValueError("Invalid treatment conversions.")

    if not 0 <= control_conversions <= control_users:
        raise ValueError("Invalid control conversions.")

    if not 0 < confidence < 1:
        raise ValueError("Confidence must be between 0 and 1.")

    p_t = treatment_conversions / treatment_users
    p_c = control_conversions / control_users

    lift = p_t - p_c
    standard_error = sqrt(
        p_t * (1 - p_t) / treatment_users
        + p_c * (1 - p_c) / control_users
    )

    z_critical = NormalDist().inv_cdf((1 + confidence) / 2)

    lower = (lift - z_critical * standard_error) * 100
    upper = (lift + z_critical * standard_error) * 100

    return {
        "absolute_lift_pp": round(lift * 100, 4),
        "confidence_level": confidence,
        "ci_lower_pp": round(lower, 4),
        "ci_upper_pp": round(upper, 4),
        "interval_includes_zero": lower <= 0 <= upper
    }


ci_result = calculate_lift_confidence_interval(
    treatment_conversions=800,
    treatment_users=10000,
    control_conversions=700,
    control_users=10000
)

print("95% confidence interval analysis:")
for key, value in ci_result.items():
    print(f"{key}: {value}")
```

```python
def classify_experiment(analysis, confidence_interval):
    lift = analysis["conversion_analysis"]["absolute_lift_pp"]
    significant = analysis["significance_analysis"]["statistically_significant"]
    lower = confidence_interval["ci_lower_pp"]
    upper = confidence_interval["ci_upper_pp"]

    if upper < 0:
        return "NEGATIVE EFFECT — investigate potential harm"

    if lower > 0 and significant and lift > 0:
        return "POSITIVE EVIDENCE — evaluate profitability before scaling"

    if lower <= 0 <= upper:
        return "INCONCLUSIVE — collect more evidence"

    if lift <= 0:
        return "NO POSITIVE OBSERVED LIFT — investigate"

    return "REVIEW — validate experiment assumptions"


decision_rows = []

for _, row in df.iterrows():
    tc = int(row["treatment_conversions"])
    tn = int(row["treatment_users"])
    cc = int(row["control_conversions"])
    cn = int(row["control_users"])

    analysis = analyze_experiment(tc, tn, cc, cn)
    ci = calculate_lift_confidence_interval(tc, tn, cc, cn)

    decision_rows.append({
        "experiment_id": row["experiment_id"],
        "absolute_lift_pp": analysis["conversion_analysis"]["absolute_lift_pp"],
        "p_value": analysis["significance_analysis"]["p_value"],
        "ci_lower_pp": ci["ci_lower_pp"],
        "ci_upper_pp": ci["ci_upper_pp"],
        "decision": classify_experiment(analysis, ci)
    })

decision_df = pd.DataFrame(decision_rows)
display(decision_df)

decision_path = OUTPUT_PATH.parent / "experiment_decisions.csv"
decision_df.to_csv(decision_path, index=False)
print("Saved:", decision_path)
```

```python
from pathlib import Path

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
API_PATH = PROJECT_DIR / "src" / "causalmark" / "api.py"

api_code = r'''
import subprocess
from pathlib import Path
from math import sqrt
from statistics import NormalDist

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field

from causalmark.analysis_engine import analyze_experiment

app = FastAPI(title="CausalMark Experiment Analysis API", version="1.1.0")

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
PYTHON = Path.home() / "qwen3-env" / "bin" / "python"
BASE_MODEL = "mlx-community/Qwen3-8B-4bit"
ADAPTER_DIR = PROJECT_DIR / "models" / "finetuned" / "causalmark-qwen3-v2"


class ExperimentRequest(BaseModel):
    experiment_id: str = "EXP001"
    treatment_conversions: int = Field(ge=0)
    treatment_users: int = Field(gt=0)
    control_conversions: int = Field(ge=0)
    control_users: int = Field(gt=0)
    alpha: float = Field(default=0.05, gt=0, lt=1)


class InterpretationRequest(ExperimentRequest):
    question: str = "Explain the experiment results and recommend the next step."


def calculate_lift_confidence_interval(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    confidence=0.95
):
    p_t = treatment_conversions / treatment_users
    p_c = control_conversions / control_users
    lift = p_t - p_c

    se = sqrt(
        p_t * (1 - p_t) / treatment_users
        + p_c * (1 - p_c) / control_users
    )

    z = NormalDist().inv_cdf((1 + confidence) / 2)
    lower = (lift - z * se) * 100
    upper = (lift + z * se) * 100

    return {
        "confidence_level": confidence,
        "ci_lower_pp": round(lower, 4),
        "ci_upper_pp": round(upper, 4),
        "interval_includes_zero": lower <= 0 <= upper
    }


def get_analysis(request):
    if request.treatment_conversions > request.treatment_users:
        raise HTTPException(400, "Treatment conversions exceed group size.")
    if request.control_conversions > request.control_users:
        raise HTTPException(400, "Control conversions exceed group size.")

    analysis = analyze_experiment(
        request.treatment_conversions,
        request.treatment_users,
        request.control_conversions,
        request.control_users,
        request.alpha
    )

    return {
        "experiment_id": request.experiment_id,
        **analysis,
        "confidence_interval": calculate_lift_confidence_interval(
            request.treatment_conversions,
            request.treatment_users,
            request.control_conversions,
            request.control_users
        )
    }


@app.get("/health")
def health():
    return {"status": "ok", "service": "CausalMark"}


@app.post("/analyze")
def analyze(request: ExperimentRequest):
    return get_analysis(request)


@app.post("/interpret")
def interpret(request: InterpretationRequest):
    analysis = get_analysis(request)

    prompt = (
        "You are CausalMark, a marketing experiment analyst. "
        "Explain the supplied results in plain language. "
        "Use the supplied calculations without changing their values. "
        "Distinguish statistical significance from profitability. "
        "Mention uncertainty and limitations. Do not invent missing costs, "
        "revenue, or business facts. Answer the user's question concisely.\n\n"
        f"User question: {request.question}\n\n"
        f"Verified experiment analysis: {analysis}"
    )

    command = [
        str(PYTHON), "-m", "mlx_lm", "generate",
        "--model", BASE_MODEL,
        "--adapter-path", str(ADAPTER_DIR),
        "--prompt", prompt,
        "--max-tokens", "250"
    ]

    try:
        result = subprocess.run(
            command,
            capture_output=True,
            text=True,
            timeout=180
        )
    except subprocess.TimeoutExpired:
        raise HTTPException(504, "Local model generation timed out.")

    if result.returncode != 0:
        raise HTTPException(
            500,
            f"Local model failed: {result.stderr[-1500:]}"
        )

    return {
        "analysis": analysis,
        "interpretation": result.stdout.strip()
    }
'''

API_PATH.write_text(api_code.strip() + "\n", encoding="utf-8")
print("Updated:", API_PATH)
```

```python
from pathlib import Path

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
API_PATH = PROJECT_DIR / "src" / "causalmark" / "api.py"

api_code = r'''
import re
import subprocess
from math import sqrt
from statistics import NormalDist
from pathlib import Path

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field

from causalmark.analysis_engine import analyze_experiment

app = FastAPI(title="CausalMark Experiment Analysis API", version="1.2.0")

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
PYTHON = Path.home() / "qwen3-env" / "bin" / "python"
BASE_MODEL = "mlx-community/Qwen3-8B-4bit"
ADAPTER_DIR = PROJECT_DIR / "models" / "finetuned" / "causalmark-qwen3-v2"


class ExperimentRequest(BaseModel):
    experiment_id: str = "EXP001"
    treatment_conversions: int = Field(ge=0)
    treatment_users: int = Field(gt=0)
    control_conversions: int = Field(ge=0)
    control_users: int = Field(gt=0)
    alpha: float = Field(default=0.05, gt=0, lt=1)


class InterpretationRequest(ExperimentRequest):
    question: str = "Explain the experiment results and recommend the next step."


def calculate_lift_confidence_interval(
    treatment_conversions,
    treatment_users,
    control_conversions,
    control_users,
    confidence=0.95
):
    p_t = treatment_conversions / treatment_users
    p_c = control_conversions / control_users
    lift = p_t - p_c

    se = sqrt(
        p_t * (1 - p_t) / treatment_users
        + p_c * (1 - p_c) / control_users
    )

    z = NormalDist().inv_cdf((1 + confidence) / 2)
    lower = (lift - z * se) * 100
    upper = (lift + z * se) * 100

    return {
        "confidence_level": confidence,
        "ci_lower_pp": round(lower, 4),
        "ci_upper_pp": round(upper, 4),
        "interval_includes_zero": lower <= 0 <= upper
    }


def get_analysis(request):
    if request.treatment_conversions > request.treatment_users:
        raise HTTPException(400, "Treatment conversions exceed group size.")
    if request.control_conversions > request.control_users:
        raise HTTPException(400, "Control conversions exceed group size.")

    analysis = analyze_experiment(
        request.treatment_conversions,
        request.treatment_users,
        request.control_conversions,
        request.control_users,
        request.alpha
    )

    return {
        "experiment_id": request.experiment_id,
        **analysis,
        "confidence_interval": calculate_lift_confidence_interval(
            request.treatment_conversions,
            request.treatment_users,
            request.control_conversions,
            request.control_users
        )
    }


def build_experiment_summary(analysis):
    conversion = analysis["conversion_analysis"]
    significance = analysis["significance_analysis"]
    ci = analysis["confidence_interval"]

    relative = conversion["relative_lift_pct"]
    relative_text = (
        f"{relative:.2f}%"
        if relative is not None else "undefined because the control rate is zero"
    )

    significance_text = (
        "statistically significant"
        if significance["statistically_significant"]
        else "not statistically significant"
    )

    return (
        f"Treatment conversion rate was {conversion['treatment_rate_pct']:.2f}%, "
        f"compared with {conversion['control_rate_pct']:.2f}% for control. "
        f"The absolute lift was {conversion['absolute_lift_pp']:.2f} "
        f"percentage points and relative lift was {relative_text}. "
        f"The result was {significance_text} at alpha={significance['alpha']:.3f} "
        f"(p={significance['p_value']:.6f}). The 95% confidence interval for "
        f"absolute lift was {ci['ci_lower_pp']:.4f} to "
        f"{ci['ci_upper_pp']:.4f} percentage points. Statistical significance "
        "does not by itself establish profitability; assess campaign cost and "
        "incremental contribution margin before making a scaling decision."
    )


def clean_llm_output(text):
    text = re.sub(r"<think>.*?</think>", "", text, flags=re.S)
    text = re.sub(r"^=+\s*$", "", text, flags=re.M)
    text = re.sub(r"^(Prompt|Generation|Peak memory):.*$", "", text, flags=re.M)
    return re.sub(r"\n{3,}", "\n\n", text).strip()


@app.get("/health")
def health():
    return {"status": "ok", "service": "CausalMark"}


@app.post("/analyze")
def analyze(request: ExperimentRequest):
    return get_analysis(request)


@app.post("/interpret")
def interpret(request: InterpretationRequest):
    analysis = get_analysis(request)
    summary = build_experiment_summary(analysis)

    prompt = (
        "You are CausalMark, a marketing experiment analyst. "
        "Explain the supplied results and answer the user's question. "
        "Use the verified numbers exactly. The absolute lift is measured in "
        "percentage points, not percent. Do not invent costs, revenue, or "
        "business facts. Statistical significance does not prove profitability. "
        "Keep the response concise.\n\n"
        f"Question: {request.question}\n\n"
        f"Verified results: {analysis}\n\n"
        f"Verified summary: {summary}"
    )

    command = [
        str(PYTHON), "-m", "mlx_lm", "generate",
        "--model", BASE_MODEL,
        "--adapter-path", str(ADAPTER_DIR),
        "--prompt", prompt,
        "--max-tokens", "250"
    ]

    try:
        generated = subprocess.run(
            command,
            capture_output=True,
            text=True,
            timeout=180
        )
    except subprocess.TimeoutExpired:
        raise HTTPException(504, "Local model generation timed out.")

    if generated.returncode != 0:
        raise HTTPException(
            500,
            f"Local model failed: {generated.stderr[-1500:]}"
        )

    return {
        "analysis": analysis,
        "summary": summary,
        "interpretation": clean_llm_output(generated.stdout)
    }
'''

API_PATH.write_text(api_code.strip() + "\n", encoding="utf-8")

root_api = PROJECT_DIR / "api.py"
root_api.write_text(API_PATH.read_text(encoding="utf-8"), encoding="utf-8")

assert root_api.read_bytes() == API_PATH.read_bytes()

print("Updated both API files.")
print("API version: 1.2.0")
print("Endpoints: /health, /analyze, /interpret")
```

```python
import subprocess
import time
import json
import urllib.request
from pathlib import Path

PROJECT_DIR = Path("/Users/naitikdonda/Desktop/CasualMark v3")
PYTHON = Path.home() / "qwen3-env" / "bin" / "python"
SRC_DIR = PROJECT_DIR / "src"

if "server_interpret" in globals() and server_interpret.poll() is None:
    server_interpret.terminate()
    server_interpret.wait(timeout=5)

server_interpret = subprocess.Popen(
    [
        str(PYTHON), "-m", "uvicorn",
        "causalmark.api:app",
        "--app-dir", str(SRC_DIR),
        "--host", "127.0.0.1",
        "--port", "8001"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

routes = None

for _ in range(20):
    try:
        with urllib.request.urlopen(
            "http://127.0.0.1:8001/openapi.json", timeout=2
        ) as response:
            schema = json.loads(response.read().decode())
            routes = list(schema.get("paths", {}).keys())
        break
    except Exception:
        time.sleep(1)

print("Registered routes:", routes)

if routes is None or "/interpret" not in routes:
    server_interpret.terminate()
    raise RuntimeError("Updated API failed to start or /interpret is missing.")

print("Updated API is running at http://127.0.0.1:8001")
```

```python

import requests

API_URL = "http://127.0.0.1:8001"

test_cases = [
    {
        "name": "Positive lift",
        "payload": {
            "experiment_id": "EXP_POSITIVE",
            "treatment_conversions": 900,
            "treatment_users": 10000,
            "control_conversions": 700,
            "control_users": 10000
        },
        "expected_lift": 2.0,
        "expected_significant": True
    },
    {
        "name": "Negative lift",
        "payload": {
            "experiment_id": "EXP_NEGATIVE",
            "treatment_conversions": 500,
            "treatment_users": 10000,
            "control_conversions": 700,
            "control_users": 10000
        },
        "expected_lift": -2.0,
        "expected_significant": True
    },
    {
        "name": "Small insignificant lift",
        "payload": {
            "experiment_id": "EXP_INSIGNIFICANT",
            "treatment_conversions": 505,
            "treatment_users": 10000,
            "control_conversions": 500,
            "control_users": 10000
        },
        "expected_lift": 0.05,
        "expected_significant": False
    }
]

results = []

for case in test_cases:
    response = requests.post(
        f"{API_URL}/analyze",
        json=case["payload"],
        timeout=30
    )

    response.raise_for_status()
    data = response.json()

    conversion = data["conversion_analysis"]
    significance = data["significance_analysis"]

    actual_lift = conversion["absolute_lift_pp"]
    actual_significance = significance["statistically_significant"]

    passed = (
        abs(actual_lift - case["expected_lift"]) < 0.001
        and actual_significance == case["expected_significant"]
    )

    results.append({
        "test": case["name"],
        "expected_lift_pp": case["expected_lift"],
        "actual_lift_pp": actual_lift,
        "expected_significant": case["expected_significant"],
        "actual_significant": actual_significance,
        "status": "PASS" if passed else "FAIL"
    })

import pandas as pd

test_results = pd.DataFrame(results)
display(test_results)

assert (test_results["status"] == "PASS").all(), "One or more regression tests failed."

print("All regression tests passed.")

```
